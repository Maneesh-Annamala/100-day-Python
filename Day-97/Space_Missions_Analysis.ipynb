## Data Cleaning - Check for Missing Values and Duplicates

Consider removing columns containing junk data. 

```python
df_clean = df_data.drop(columns=['Unnamed: 0.1', 'Unnamed: 0']).copy()

df_clean['Price'] = (df_clean['Price'].astype(str).str.replace(',', '', regex=False).replace('nan', np.nan))
df_clean['Price'] = pd.to_numeric(df_clean['Price'], errors='coerce')

df_clean['Date'] = pd.to_datetime(df_clean['Date'], utc=True, errors='coerce', format='mixed')
df_clean['Year'] = df_clean['Date'].dt.year
df_clean['Month_Num'] = df_clean['Date'].dt.month
df_clean['Month'] = df_clean['Date'].dt.month_name()
df_clean['Month_Year'] = df_clean['Date'].dt.to_period('M').dt.to_timestamp()

df_clean['Rocket_Status'] = df_clean['Rocket_Status'].str.replace('Status', '', regex=False)
df_clean.head()
```

# Use a Choropleth Map to Show the Number of Launches by Country

* Create a choropleth map using [the plotly documentation](https://plotly.com/python/choropleth-maps/)
* Experiment with [plotly's available colours](https://plotly.com/python/builtin-colorscales/). I quite like the sequential colour `matter` on this map. 
* You'll need to extract a `country` feature as well as change the country names that no longer exist.

Wrangle the Country Names

You'll need to use a 3 letter country code for each country. You might have to change some country names.

* Russia is the Russian Federation
* New Mexico should be USA
* Yellow Sea refers to China
* Shahrud Missile Test Site should be Iran
* Pacific Missile Range Facility should be USA
* Barents Sea should be Russian Federation
* Gran Canaria should be USA


You can use the iso3166 package to convert the country names to Alpha3 format.

```python
country_replacements = {
    'Russia': 'Russian Federation',
    'New Mexico': 'USA',
    'Yellow Sea': 'China',
    'Shahrud Missile Test Site': 'Iran',
    'Pacific Missile Range Facility': 'USA',
    'Barents Sea': 'Russian Federation',
    'Gran Canaria': 'USA',
}

alpha3_lookup = {
    'Australia': 'AUS',
    'Brazil': 'BRA',
    'China': 'CHN',
    'France': 'FRA',
    'India': 'IND',
    'Iran': 'IRN',
    'Israel': 'ISR',
    'Japan': 'JPN',
    'Kazakhstan': 'KAZ',
    'Kenya': 'KEN',
    'New Zealand': 'NZL',
    'North Korea': 'PRK',
    'Pacific Ocean': 'PCE',
    'Russian Federation': 'RUS',
    'South Korea': 'KOR',
    'USA': 'USA',
}

def country_to_alpha3(country_name):
    if country_name in alpha3_lookup:
        return alpha3_lookup[country_name]
    if countries is not None:
        try:
            return countries.get(country_name).alpha3
        except KeyError:
            return np.nan
    return np.nan

df_clean['Country'] = df_clean['Location'].str.split(',').str[-1].str.strip()
df_clean['Country'] = df_clean['Country'].replace(country_replacements)
df_clean['Country_Code'] = df_clean['Country'].apply(country_to_alpha3)

df_clean[['Location', 'Country', 'Country_Code']].drop_duplicates().head(20)
```

```python
launches_by_country = (df_clean.groupby(['Country', 'Country_Code'], dropna=False).size().reset_index(name='Launches').sort_values('Launches', ascending=False))
display(launches_by_country)

fig = px.choropleth(
    launches_by_country[launches_by_country['Country_Code'] != 'PCE'],
    locations='Country_Code',
    color='Launches',
    hover_name='Country',
    color_continuous_scale='matter',
    title='Number of launches by country',
)
fig.show()
```

# Cold War Space Race: USA vs USSR

The cold war lasted from the start of the dataset up until 1991. 

```python
cold_war = df_clean[df_clean['Year'].between(1957, 1991)].copy()
cold_war['Superpower'] = np.select(
    [
        cold_war['Country'].eq('USA'),
        cold_war['Country'].isin(['Russian Federation', 'Kazakhstan']),
    ],
    ['USA', 'USSR'],
    default='Other',
)
cold_war_superpowers = cold_war[cold_war['Superpower'].isin(['USA', 'USSR'])].copy()
cold_war_superpowers[['Date', 'Country', 'Organisation', 'Mission_Status', 'Superpower']].head()
```

```python
cold_war_summary = (
    cold_war_superpowers.groupby('Superpower')
    .agg(Launches=('Mission_Status', 'size'), Failures=('Mission_Status', lambda s: s.str.contains('Failure').sum()))
)
cold_war_summary['Failure_Rate'] = cold_war_summary['Failures'] / cold_war_summary['Launches']
cold_war_summary
```

## Chart the Percentage of Failures over Time

Did failures go up or down over time? Did the countries get better at minimising risk and improving their chances of success over time? 

```python
cold_war_status_yearly = (
    cold_war_superpowers.assign(Is_Failure=cold_war_superpowers['Mission_Status'].str.contains('Failure', case=False, na=False))
    .groupby(['Year', 'Superpower'])
    .agg(Launches=('Is_Failure', 'size'), Failures=('Is_Failure', 'sum'))
    .reset_index()
)
cold_war_status_yearly['Failure_Rate'] = cold_war_status_yearly['Failures'] / cold_war_status_yearly['Launches']
cold_war_status_yearly.head()
```

```python
fig = px.line(
    cold_war_status_yearly,
    x='Year',
    y='Failure_Rate',
    color='Superpower',
    markers=True,
    title='Cold War failure percentage over time',
    labels={'Failure_Rate': 'Failure rate'},
)
fig.update_yaxes(tickformat='.0%')
fig.show()
```

```python
early = cold_war_status_yearly[cold_war_status_yearly['Year'].between(1957, 1966)].groupby('Superpower')['Failure_Rate'].mean()
late = cold_war_status_yearly[cold_war_status_yearly['Year'].between(1982, 1991)].groupby('Superpower')['Failure_Rate'].mean()
comparison = pd.DataFrame({'1957-1966 average': early, '1982-1991 average': late})
comparison['Change'] = comparison['1982-1991 average'] - comparison['1957-1966 average']
display(comparison)
print('Failure rates generally fell from the first decade of the Space Race to the final decade of the Cold War.')
```
